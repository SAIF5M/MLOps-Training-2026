```python
import pandas as pd
import numpy as np
import os
import joblib

train_df = pd.read_csv('artifacts/train.csv')
val_df = pd.read_csv('artifacts/validation.csv')
test_df = pd.read_csv('artifacts/test.csv')

print("Original shapes - Train:", train_df.shape, "Val:", val_df.shape, "Test:", test_df.shape)

median_price = train_df['price'].median()

train_df['price'] = train_df['price'].fillna(median_price)
val_df['price'] = val_df['price'].fillna(median_price)
test_df['price'] = test_df['price'].fillna(median_price)

print("Missing values in price after imputation (Train):", train_df['price'].isnull().sum())

os.makedirs('artifacts', exist_ok=True)
train_df.to_csv('artifacts/train_engineered.csv', index=False)
val_df.to_csv('artifacts/val_engineered.csv', index=False)
test_df.to_csv('artifacts/test_engineered.csv', index=False)

print("Engineered datasets saved successfully in artifacts!")
```

```python
numeric_cols_to_fix = ['freight_value', 'total_items_count', 'payment_value', 'payment_installments']

for col in numeric_cols_to_fix:
    if col in train_df.columns:
        train_median = train_df[col].median()
        
        train_df[col] = train_df[col].fillna(train_median)
        val_df[col] = val_df[col].fillna(train_median)
        test_df[col] = test_df[col].fillna(train_median)
        print(f"Imputed missing values for: {col} using train median = {train_median}")

train_df.to_csv('artifacts/train_engineered.csv', index=False)
val_df.to_csv('artifacts/val_engineered.csv', index=False)
test_df.to_csv('artifacts/test_engineered.csv', index=False)

print("\nAll missing values handled and engineered datasets updated successfully in artifacts!")
```

```python
import json

feature_list = ['price', 'freight_value', 'total_items_count', 'payment_value', 'payment_installments']

os.makedirs('artifacts', exist_ok=True)
with open('artifacts/feature_list.json', 'w') as f:
    json.dump(feature_list, f)

print("Feature list saved as an artifact successfully!")
```

```python
from sklearn.impute import SimpleImputer
import joblib

num_imputer = SimpleImputer(strategy='median')
num_imputer.fit(train_df[['price', 'freight_value', 'total_items_count', 'payment_value', 'payment_installments']])

joblib.dump(num_imputer, 'artifacts/fitted_num_imputer.pkl')
print("Fitted transformer (imputer) saved successfully as an artifact!")
```
